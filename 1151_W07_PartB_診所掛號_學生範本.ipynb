## 4. AI 的工作

* 對象只能是：`內科`、`外科`、`牙科`、`未知`


* 無法判斷時輸出：`未知,未知`


* AI 不可以做的事：

4. 不可輸出除了 `動作,對象` 以外的任何文字、解釋或標點符號。

````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型
# ============================================================
!pip install -q groq gradio

import sqlite3
import pandas as pd
from IPython.display import display
from google.colab import userdata
from groq import Groq
import gradio as gr

# 🔴 金鑰只能從左側「🔑 密鑰（Secrets）」讀取，絕對不可以寫成 api_key="gsk_..."
#    寫死的金鑰一旦存到 GitHub，就等於公開給全世界使用。
#    也不要 print 出金鑰，因為輸出結果也會一起存到 GitHub。
try:
    client = Groq(api_key=userdata.get('GROQ_API_KEY'))
except Exception as e:
    print("❌ 讀不到金鑰。請檢查：")
    print("   1. 左側 🔑 是否新增了名稱為 GROQ_API_KEY 的密鑰（大小寫要完全一樣）")
    print("   2. 該密鑰的「筆記本存取權」開關是否已打開")
    raise

# 模型名稱集中放在這裡；日後模型更換或下架，只要改這一行
MODEL = "openai/gpt-oss-120b"
print("✅ Groq 金鑰讀取成功，使用模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# ============================================================
import sqlite3

DB_PATH = "clinic.db"

def init_db():
    conn = None
    try:
        conn = sqlite3.connect(DB_PATH)
        cursor = conn.cursor()

        # 1. 每次執行皆 DROP TABLE IF EXISTS 重建
        cursor.execute("DROP TABLE IF EXISTS appointments")
        cursor.execute("DROP TABLE IF EXISTS departments")

        # 2. 資料表 1：各科名額表（最多 2 張表規範）
        cursor.execute("""
            CREATE TABLE departments (
                name TEXT PRIMARY KEY,
                total_quota INTEGER NOT NULL,
                booked_count INTEGER NOT NULL DEFAULT 0
            )
        """)

        # 3. 資料表 2：掛號紀錄表
        cursor.execute("""
            CREATE TABLE appointments (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                department TEXT NOT NULL,
                created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
            )
        """)

        # 4. 寫入初始資料（僅限內科、外科、牙科，每科預設 5 名）
        initial_depts = [
            ("內科", 5, 0),
            ("外科", 5, 0),
            ("牙科", 5, 0)
        ]

        # 全程使用 ? 參數化查詢防範 SQL 注入
        cursor.executemany(
            "INSERT INTO departments (name, total_quota, booked_count) VALUES (?, ?, ?)",
            initial_depts
        )

        conn.commit()
    finally:
        # 使用 try/finally 確保資料庫連線正確關閉
        if conn:
            conn.close()

# 執行初始化
init_db()
print("資料庫初始化完成（已重建 departments 與 appointments 資料表）。")
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# ============================================================
import sqlite3

DB_PATH = "clinic.db"

def show_tables():
    """查詢並顯示 departments 與 appointments 資料表的完整內容"""
    conn = None
    try:
        conn = sqlite3.connect(DB_PATH)
        cursor = conn.cursor()

        # 1. 顯示 departments 表內容
        print("=== departments 資料表 ===")
        cursor.execute("SELECT name, total_quota, booked_count FROM departments")
        dept_rows = cursor.fetchall()
        print(f"{'科別':<6} | {'總名額':<6} | {'已掛號數':<6}")
        print("-" * 30)
        for name, total, booked in dept_rows:
            print(f"{name:<6} | {total:<8} | {booked:<8}")

        # 2. 顯示 appointments 表內容
        print("\n=== appointments 資料表 ===")
        cursor.execute("SELECT id, department, created_at FROM appointments")
        appt_rows = cursor.fetchall()
        if not appt_rows:
            print("（目前無掛號紀錄）")
        else:
            print(f"{'編號':<6} | {'科別':<6} | {'掛號時間':<20}")
            print("-" * 38)
            for appt_id, dept, created_at in appt_rows:
                print(f"{appt_id:<8} | {dept:<6} | {created_at}")

    except Exception as e:
        print(f"查詢資料表時發生錯誤：{e}")
    finally:
        # 使用 try/finally 確保連線正確關閉
        if conn:
            conn.close()

# 呼叫一次函式以印出目前資料庫狀態
show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 科別, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# ============================================================
from groq import Groq
from google.colab import userdata

# 1. 安全讀取金鑰（禁止 print 金鑰內容）
GROQ_API_KEY = userdata.get("GROQ_API_KEY")
if not GROQ_API_KEY:
    raise ValueError("請先在 Colab Secrets 中設定 GROQ_API_KEY！")

groq_client = Groq(api_key=GROQ_API_KEY)

# 2. 定義系統提示詞（包含安全紅線與固定輸出格式規範）
SYSTEM_PROMPT = """你是一個診所掛號意圖解析器。你的任務是將使用者的語句分類為「動作」與「對象」，並輸出為格式：`動作,對象`。

【業務範圍】
- 動作僅限：掛號、取消掛號、查詢名額、未知
- 對象僅限：內科、外科、牙科、未知

【絕對核心紅線（違者嚴重處置）】
1. 你絕對不是醫生，嚴禁提供任何醫療建議、處方或衛教資訊！
2. 絕對不可根據症狀推薦科別！
3. 若使用者輸入包含任何病症、症狀描述（如：肚子痛、頭痛、牙痛、發燒、想吐等），或詢問「看哪科」、「該掛哪科」，一律輸出：`未知,未知`。
4. 若無法明確判斷動作，動作標記為「未知」；若未提及內科、外科、牙科，對象標記為「未知」。
5. 請只輸出「動作,對象」，不要包含任何額外說明、問候或標點符號。

【範例】
- 「我想掛內科」 -> 掛號,內科
- 「請問外科還有名額嗎？」 -> 查詢名額,外科
- 「退掛牙科」 -> 取消掛號,牙科
- 「我肚子很痛」 -> 未知,未知
- 「牙齒痛看哪科」 -> 未知,未知
- 「我要看眼科」 -> 掛號,未知
- 「你好」 -> 未知,未知
"""

# 白名單定義（不可直接相信 AI，後續進行驗證校對）
ALLOWED_ACTIONS = {"掛號", "取消掛號", "查詢名額"}
ALLOWED_DEPTS = {"內科", "外科", "牙科"}

def parse_user_intent(user_input: str) -> tuple[str, str, str]:
    """
    呼叫 Groq 模型解讀意圖，並回傳 (動作, 科別, 原始回覆)
    透過白名單機制防範 AI 幻覺與非預期回傳
    """
    raw_response = ""
    try:
        completion = groq_client.chat.completions.create(
            model="llama-3.3-70b-versatile",
            messages=[
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": user_input.strip()}
            ],
            temperature=0.0,
            max_tokens=20
        )
        raw_response = completion.choices[0].message.content.strip()
    except Exception as e:
        raw_response = f"API_ERROR: {str(e)}"
        return "未知", "未知", raw_response

    # 解析 AI 輸出並去除多餘符號（如反引號、空白）
    cleaned = raw_response.replace("`", "").strip()
    parts = [p.strip() for p in cleaned.split(",")]

    if len(parts) == 2:
        action, dept = parts[0], parts[1]
    else:
        action, dept = "未知", "未知"

    # 白名單驗證：若不在允許清單內，強制退回「未知」
    if action not in ALLOWED_ACTIONS:
        action = "未知"
    if dept not in ALLOWED_DEPTS:
        dept = "未知"

    return action, dept, raw_response

# 快速單元測試驗證
test_action, test_dept, test_raw = parse_user_intent("我想掛內科")
print(f"測試輸出 -> 動作: {test_action} | 科別: {test_dept} | 原始回覆: {test_raw}")
````

````python
# ============================================================
# 第 5 格：測試清單
# 提醒：至少 8 句，要與規格書第 7 節一致
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# ============================================================
# 定義測試語句清單（完全對應規格書第 7 節之正常、錯誤、邊界情境）
test_sentences = [
    "我想掛內科",                           # 1. 正常：掛號
    "請問外科還有名額嗎？",                 # 2. 正常：查詢名額
    "牙科幫我退掛",                         # 3. 正常：取消掛號
    "我肚子痛看哪一科？",                   # 4. 錯誤（安全紅線）：症狀詢問
    "頭好痛想拿止痛藥",                     # 5. 錯誤（安全紅線）：症狀描述
    "幫我掛眼科門診",                       # 6. 錯誤（不支援科別）：非白名單科別
    "今天天氣真好",                         # 7. 錯誤（無關輸入）：無法判斷
    "我要取消掛號",                         # 8. 邊界（缺少科別）：未指定對象
    "我要掛內科"                            # 9. 邊界（額滿掛號）：多次掛號/重複測試
]

print(f"=== 開始執行測試清單（共 {len(test_sentences)} 句，與規格書第 7 節一致）===\n")

# 依序使用第 4 格的 parse_user_intent 函式進行意圖解讀測試
for idx, sentence in enumerate(test_sentences, 1):
    action, dept, raw_output = parse_user_intent(sentence)
    print(f"測試 #{idx}: 「{sentence}」")
    print(f"  ├─ AI 原始回覆 : {raw_output}")
    print(f"  └─ 驗證後結果 : 動作 =【{action}】, 科別 =【{dept}】")
    print("-" * 55)
````

````python
# ============================================================
# 第 6 格：功能函式（掛號、取消、查詢）
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# ============================================================
import sqlite3

DB_PATH = "clinic.db"

def query_quota(dept: str) -> str:
    """查詢指定科別的剩餘名額狀態"""
    conn = None
    try:
        conn = sqlite3.connect(DB_PATH)
        cursor = conn.cursor()

        # 使用 ? 參數化查詢
        cursor.execute("SELECT total_quota, booked_count FROM departments WHERE name = ?", (dept,))
        row = cursor.fetchone()

        # 嚴格檢查 fetchone() 是否為 None
        if row is None:
            return f"查無【{dept}】之資訊。"

        total_quota, booked_count = row
        remaining = total_quota - booked_count
        return f"【{dept}】目前掛號狀況：總名額 {total_quota} 位，已掛號 {booked_count} 位，剩餘名額 {remaining} 位。"
    finally:
        # 使用 try/finally 確保連線關閉
        if conn:
            conn.close()


def book_appointment(dept: str) -> str:
    """辦理指定科別掛號"""
    conn = None
    try:
        conn = sqlite3.connect(DB_PATH)
        cursor = conn.cursor()

        cursor.execute("SELECT total_quota, booked_count FROM departments WHERE name = ?", (dept,))
        row = cursor.fetchone()

        if row is None:
            return f"查無【{dept}】之資訊。"

        total_quota, booked_count = row
        remaining = total_quota - booked_count

        # 檢查名額是否已滿
        if remaining <= 0:
            return f"抱歉，今日【{dept}】名額已滿，無法辦理掛號。"

        # 更新已掛號人數並新增掛號紀錄
        cursor.execute("UPDATE departments SET booked_count = booked_count + 1 WHERE name = ?", (dept,))
        cursor.execute("INSERT INTO appointments (department) VALUES (?)", (dept,))
        conn.commit()

        return f"掛號成功！已為您完成【{dept}】掛號。目前該科剩餘名額：{remaining - 1} 位。"
    finally:
        if conn:
            conn.close()


def cancel_appointment(dept: str) -> str:
    """取消指定科別掛號（退掛）"""
    conn = None
    try:
        conn = sqlite3.connect(DB_PATH)
        cursor = conn.cursor()

        cursor.execute("SELECT total_quota, booked_count FROM departments WHERE name = ?", (dept,))
        row = cursor.fetchone()

        if row is None:
            return f"查無【{dept}】之資訊。"

        total_quota, booked_count = row
        remaining = total_quota - booked_count

        # 檢查是否有可取消的紀錄
        if booked_count <= 0:
            return f"目前【{dept}】尚無任何掛號紀錄，無法執行取消。"

        # 扣減掛號數並移除最新一筆掛號紀錄
        cursor.execute("UPDATE departments SET booked_count = booked_count - 1 WHERE name = ?", (dept,))
        cursor.execute(
            "DELETE FROM appointments WHERE id = (SELECT id FROM appointments WHERE department = ? ORDER BY id DESC LIMIT 1)",
            (dept,)
        )
        conn.commit()

        return f"取消掛號成功！已為您取消【{dept}】掛號。目前該科剩餘名額：{remaining + 1} 位。"
    finally:
        if conn:
            conn.close()


def execute_action(action: str, dept: str) -> str:
    """整合函式：依據動作與科別分派執行，並處理未知或例外情況"""
    # 白名單與未知檢查（安全紅線及格式防護）
    if action == "未知" or dept == "未知":
        return (
            "抱歉，系統無法判斷您的需求。\n"
            "※ 本診所僅提供【內科】、【外科】、【牙科】之服務。\n"
            "※ 本系統不提供醫療診斷與症狀推薦科別服務。若身體不適，請洽門診櫃台諮詢，並請明確告知掛號科別。"
        )

    if action == "查詢名額":
        return query_quota(dept)
    elif action == "掛號":
        return book_appointment(dept)
    elif action == "取消掛號":
        return cancel_appointment(dept)
    else:
        return "系統無法識別該操作指令。"

# 快速驗證呼叫
print(execute_action("查詢名額", "內科"))
````

````python
# ============================================================
# 第 7 格：主流程與完整流程測試
# 提醒：最後呼叫 show_tables() 確認資料庫有變化
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# ============================================================
def process_message(user_input: str):
    """
    主流程函式：
    1. 透過 parse_user_intent 取得 (動作, 科別, 原始回覆)
    2. 透過 execute_action 執行資料庫操作或例外回應
    3. 印出完整追蹤資訊
    """
    # 1. 意圖解析（不可直接相信 AI，內部已做白名單驗證）
    action, dept, raw_output = parse_user_intent(user_input)

    # 2. 執行業務邏輯與資料庫更新
    reply = execute_action(action, dept)

    # 3. 顯示執行歷程
    print(f"使用者輸入 : 「{user_input}」")
    print(f"AI 原始回覆: {raw_output}")
    print(f"解析判斷   : 動作 =【{action}】, 科別 =【{dept}】")
    print(f"系統回覆   :\n{reply}")
    print("-" * 60)
    return reply


# ------------------------------------------------------------
# 完整流程自動化測試（使用第 5 格定義之 test_sentences）
# ------------------------------------------------------------
print(">>> 開始執行主流程完整測試 <<<\n")

for idx, sentence in enumerate(test_sentences, 1):
    print(f"【測試案例 #{idx}】")
    process_message(sentence)

# ------------------------------------------------------------
# 依提醒：最後呼叫 show_tables() 檢查資料庫狀態與變化
# ------------------------------------------------------------
print("\n>>> 測試完成，檢視資料庫最終變化狀態 <<<")
show_tables()
````

````python
# ============================================================
# 第 8 格：Gradio 網頁
# 提醒：放在最後一格；GitHub 上看不到網頁，批改看的是第 5、7 格的結果
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# ============================================================
import gradio as gr

def clinic_chat(message: str, history):
    """
    接收使用者訊息，依序透過第 4 格與第 6 格的函式處理：
    1. parse_user_intent: 呼叫 AI 解析意圖並回傳 (動作, 科別, 原始回覆)
    2. execute_action: 執行業務邏輯、資料庫操作與白名單例外處理
    """
    if not message.strip():
        return "請輸入您的需求（如：我想掛內科、查詢外科名額）。"

    # 1. 意圖解讀（包含白名單校驗與安全紅線過濾）
    action, dept, raw_output = parse_user_intent(message)

    # 2. 執行資料庫動作與回應生成
    reply = execute_action(action, dept)

    # 3. 組合展示 AI 判斷與系統正式回覆
    response = (
        f"【AI 原始解析】：`{raw_output}`\n"
        f"【意圖判定】：動作 = **{action}** / 科別 = **{dept}**\n\n"
        f"【系統回覆】：\n{reply}"
    )
    return response

# 建立 Gradio Chat 介面
demo = gr.ChatInterface(
    fn=clinic_chat,
    title="🏥 診所智能掛號系統",
    description=(
        "歡迎使用掛號小幫手！本診所僅提供【內科】、【外科】、【牙科】之服務。\n"
        "支援功能：**掛號**、**取消掛號**、**查詢名額**。\n"
        "⚠️ 本系統依醫療安全規範，嚴禁提供醫療診斷與症狀推薦科別。"
    ),
    examples=[
        "我想掛內科",
        "請問外科還有名額嗎？",
        "幫我取消牙科掛號",
        "我肚子痛該看哪一科？",
        "幫我掛眼科門診"
    ]
)

# 啟動 Gradio（若在 Colab 執行可加上 share=True 取得公開連結）
demo.launch(share=True)
````
